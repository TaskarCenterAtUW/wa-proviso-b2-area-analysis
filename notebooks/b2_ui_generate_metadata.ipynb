```python
import os
import json

county_name = 'yakima'

# Get the current metadata file
metadata_file_path = f'../ui-union/metadata/{county_name}.json'

# Get the new boundary path
new_boundary_path = f'../ui-union/output_boundaries/{county_name}_ui_with_pop_centers.geojson'

# Check if both files exist
if not os.path.exists(metadata_file_path) or not os.path.exists(new_boundary_path):
    print('Required files donot exist')
else:
    print('Metadata and the new boundary files exist')

# existing_metadata = 
with open(metadata_file_path) as f:
    existing_metadata = json.load(f)
existing_version = existing_metadata['dataset_detail']['version']    
name = existing_metadata['dataset_detail']['name']


```

```python
def get_new_version(old_version):
    # split the version into major, minor
    major, minor = old_version.split('.')
    # increment the patch version
    minor = str(int(minor) + 1)
    # If minor is more than 9, increment the major version
    if int(minor) > 9:
        major = str(int(major) + 1)
        minor = '0'
    # return the new version
    return f'{major}.{minor}'


with open(new_boundary_path,'r') as f:
    new_boundary_data = json.load(f)

print(new_boundary_data)

new_metadata = existing_metadata.copy()
new_metadata['dataset_detail']['dataset_area'] = new_boundary_data
new_version = get_new_version(existing_version)
new_metadata['dataset_detail']['version'] = new_version
county_name_camel = county_name.capitalize()
new_metadata['dataset_detail']['description'] = f'OSW v0.3 compliant dataset for {county_name_camel} Unincorporated Area'
# Get the custom metadata
custom_metadata = new_metadata['dataset_detail']['custom_metadata']
if not custom_metadata:
    custom_metadata = {}

# Get the list of population centers
first_feature = new_boundary_data['features'][0]
population_centers = first_feature['properties']['pop_centers']
custom_metadata['pop_centers'] = population_centers
new_metadata['dataset_detail']['custom_metadata'] = custom_metadata

release_notes = new_metadata['dataset_summary']['release_notes']
release_notes = 'Added B2 Population centers '+release_notes
new_metadata['dataset_summary']['release_notes'] = release_notes

print(new_metadata['dataset_detail']['custom_metadata'] )

new_metadata_file_path = f'../ui-union/metadata_new/{county_name}.json'
with open(new_metadata_file_path,'w') as f:
    json.dump(new_metadata,f)
```
