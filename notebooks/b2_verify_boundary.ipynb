```python
# Code to verify the boundary
import geopandas as gpd
county_name = 'king'
boundary_geojson = f'../ui-union/output_boundaries/{county_name}_ui_with_pop_centers.geojson'
# boundary_geojson = '../ui-union/cleaned_boundaries/king_boundary.geojson'

boundary_gdf = gpd.read_file(boundary_geojson)


from shapely.validation import explain_validity
exploded_gdf = boundary_gdf.explode()
def is_valid(shape):
    try:
        shape.is_valid
        if not shape.is_valid:
            print(explain_validity(shape))
        return shape.is_valid
    except Exception:
        return False
exploded_gdf['is_valid'] = exploded_gdf['geometry'].apply(is_valid)
# check count of invalid polygons
invalid_polygons_gdf = exploded_gdf[exploded_gdf['is_valid'] == False]
print(len(invalid_polygons_gdf))
```

```python
from shapely import set_precision

boundary_gdf['geometry'] = boundary_gdf['geometry'].apply(lambda geom: set_precision(geom, grid_size=1e-7))
boundary_gdf['geometry'] = boundary_gdf.simplify(tolerance=0.001,preserve_topology=True)
# union_geometry.simplify(0)
#Use this to reduce precision
boundary_gdf.to_file(boundary_geojson,driver='GeoJSON')
```
