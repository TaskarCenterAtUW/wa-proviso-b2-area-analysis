```python
import geopandas as gpd

b2_coverage_file = '../datasets/b2_coverage_pop_centers_reduced_precision.geojson'
trail_heads_file = '../datasets/trailheads_not_in_b1.geojson'
```

```python
b2_gdf = gpd.read_file(b2_coverage_file)
trail_heads_gdf = gpd.read_file(trail_heads_file)
```

```python
trail_heads_gdf['In_B1_Area']
```

```python
# Get the union of b2_gdf
b2_gdf_union = b2_gdf.union_all()
```

```python
import shapely

def isInB2(geom):
    return geom.within(b2_gdf_union)

trail_heads_gdf['In_B2_Area'] = trail_heads_gdf['geometry'].apply(isInB2)
```

```python
trail_heads_gdf['In_B2_Area'].value_counts()
```

```python
trail_heads_not_in_b2 = trail_heads_gdf[trail_heads_gdf['In_B2_Area'] == False]
trail_heads_not_in_b2['county'].value_counts()
```

```python
trail_heads_not_in_b2.to_file('../datasets/trailheads_not_in_b2.geojson',driver='GeoJSON')
```

```python
# Get the buffered tile with 500m on each side
trail_heads_not_in_b2 = trail_heads_not_in_b2.to_crs(epsg=3857)
```

```python
trail_heads_not_in_b2['geometry'] = trail_heads_not_in_b2.geometry.buffer(250,cap_style=3)
```

```python
trail_heads_not_in_b2 = trail_heads_not_in_b2.to_crs(epsg=4326)
```

```python
trail_heads_not_in_b2.to_file('../datasets/trailheads_not_in_b2_buffered.geojson',driver='GeoJSON')
```
